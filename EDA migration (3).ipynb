```python
#Visualize the data
```

```python
plt.figure(figsize=(10, 5))

df["destination_country"].value_counts().head(10).plot(kind="bar")

plt.title("Top 10 Destination Countries")
plt.xlabel("Destination country")
plt.ylabel("Number of Students")
plt.legend()
```

```python
import matplotlib.pyplot as plt
df[df['destination_country'] == 'UAE']['enrollment_reason'].value_counts().plot(kind='bar')

plt.title('Reasons for Choosing UAE')
plt.xlabel('Reason')
plt.ylabel('Number of Students')
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(10, 5))

df["origin_country"].value_counts().head(10).plot(kind="bar")

plt.title("Top 10 Destination Countries")
plt.xlabel("origin_country")
plt.ylabel("Number of Students")
plt.legend()
```

```python

df['placement_status'].value_counts().plot(kind="pie",autopct="%1.1f%%")
plt.title("placed vs not placed")
plt.ylabel(" ")
plt.legend()
plt.show()
```

```python
df.groupby("placement_country")['starting_salary_usd'].mean().plot(kind="bar")
plt.title("placed country starting salary")
plt.xlabel("placement country")
plt.ylabel("starting salary")
plt.xticks(rotation=45)
plt.legend()
plt.show()
```

```python
df[df["placement_country"]!="unknown"].groupby("placement_country")["starting_salary_usd"].mean().plot(kind="bar")
plt.title("placed country starting salary")
plt.xlabel("placement country")
plt.ylabel("starting salary")
plt.xticks(rotation=45)
plt.legend()
plt.show()
```

```python
df.groupby("field_of_study")["starting_salary_usd"].mean().plot(kind="bar")
plt.title("field of salary")
plt.xlabel("field_of_study")
plt.ylabel("starting_salary_usd")
plt.xticks(rotation=45)
plt.legend()
plt.show()
```

```python
df.groupby('origin_country')['scholarship_received'].count().plot(kind="bar")
plt.title('Scholarships by Origin Country')
plt.xlabel('Origin Country')
plt.ylabel('scholarship_received')
plt.xticks(rotation=45)
plt.show()
```

```python

sns.countplot(x=df['origin_country'],hue=df['scholarship_received'])
plt.title('Scholarship Status by Origin Country')
plt.xlabel('Origin Country')
plt.ylabel('Number of Students')
plt.xticks(rotation=45)
plt.show()
```
